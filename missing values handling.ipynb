```python
import numpy as np       #handles the mathematical operations
import pandas as pd        #allow values to be represented in tearms of tables 
# Sample dataset
data = {
"Age": [25, 28, np.nan, 32, 100, 29],
"Salary": [50000, np.nan, 45000, 52000, 51000, 700000],       #sample dataset
"Department": ["HR", "IT", "IT", np.nan, "Finance", "HR"]
}
df = pd.DataFrame(data)     #helps the values in the dataset to be saved in terms of tables using the function dataframe from the lib pd
print("Original Data:")      #printing the datas in tables before pre processing 
print(df) 
# -----------------------------
# Handling Missing Data
# -----------------------------
df["Age"].fillna(df["Age"].mean(), inplace=True)           #fillna -to fill the values withput data....mean bc its  it has an small outliner value so easy to predict the missing value
df["Salary"].fillna(df["Salary"].median(), inplace=True)   #median bc there exist an huge outliner so taking mean would produce too much biased value...so taking the middle values while ignoring the outliner 
df["Department"].fillna(df["Department"].mode()[0], inplace=True)      #mode bc we cant calculate median or mean without numbers ,so it just takes the value that is been repeated
# -----------------------------
# Outlier Treatment (IQR method)
# -----------------------------
Q1 = df["Salary"].quantile(0.25)   #imagine you line up all your employees in order of salary, from the poorest to the richest.
Q3 = df["Salary"].quantile(0.75)     #You walk 25% of the way down the line. The person standing there marks the "Lower Quartile."
IQR = Q3 - Q1                  #You walk 75% of the way down the line. The person standing there marks the "Upper Quartile."
lower = Q1 - 1.5 * IQR           #calculating the reminding 50 percent represented as IQR
upper = Q3 + 1.5 * IQR          #The "Lower Fence." Any salary lower than this is suspiciously low.
# Cap outliers                              #The "Upper Fence." Any salary higher than this is suspiciously high (like the 700k value).
df["Salary"] = np.where(df["Salary"] > upper, upper,        #This line caps all salary values to stay within your calculated range, replacing any extreme outliers with the maximum or minimum allowed values.
np.where(df["Salary"] < lower, lower, df["Salary"]))
# -----------------------------
# Encoding Categorical Data
# -----------------------------
df_encoded = pd.get_dummies(df, columns=["Department"])   #one hot encoding 
print("\nProcessed Data:")  #converting catogorical value to numerical value
print(df_encoded)
```
